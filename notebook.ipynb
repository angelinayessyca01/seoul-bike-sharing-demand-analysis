### 4. Mengecek apakah terdapat perbedaan yang signifikan pada rata - rata jumlah penyewaan sepeda antar musim (Winter, Spring, Summer dan Autumn)

Setelah pola penyewaan sepeda berdasarkan jam dan musim telah dianalisis melalui visualisasi Line Chart di no 2. Maka akan dilakukan pengujian ANOVA untuk mengetahui apakah perbedaan rata - rata jumlah penyewaan antar musim tersebut signifikan secara statistik atau hanya disebabkan oleh variasi acak.

#### Analisis Inferensial

H0 : Tidak terdapat perbedaan yang signifikan pada rata - rata jumlah penyewaan sepeda antar musim (Winter, Spring, Summer dan Autumn)

H1 : Terdapat perbedaan yang signifikan pada rata - rata jumlah penyewaan sepeda antar musim (Winter, Spring, Summer dan Autumn)

````python
# Membandingkan antar musim menggunakan One-Way ANOVA
winter = df_functioning[
    df_functioning["Seasons"] == "Winter"
]["Rented Bike Count"]

spring = df_functioning[
    df_functioning["Seasons"] == "Spring"
]["Rented Bike Count"]

summer = df_functioning[
    df_functioning["Seasons"] == "Summer"
]["Rented Bike Count"]

autumn = df_functioning[
    df_functioning["Seasons"] == "Autumn"
]["Rented Bike Count"]
````

````python
winter
````

````python
spring
````

````python
summer
````

````python
autumn
````

````python
# Melihat rata - rata antar musim
rata2_winter = winter.mean()
rata2_spring = spring.mean()
rata2_summer = summer.mean()
rata2_autumn = autumn.mean()

print("Rata - Rata Winter :", rata2_winter)
print("Rata - Rata Spring :", rata2_spring)
print("Rata - Rata Summer :", rata2_summer)
print("Rata - Rata Autumn :", rata2_autumn)
````

````python
# Pengujian menggunakan ANOVA
uji_anova = stats.f_oneway(
    winter,
    spring,
    summer,
    autumn
)

print("F-statistic :", uji_anova.statistic)
print("P-value :", uji_anova.pvalue)
````

Data Insight : Berdasarkan hasil analisis didapatkan hasil bahwa rata - rata penyewaan sepeda tertinggi itu pada `Musim Panas (Summer)` dengan penyewaan sepeda sekitar `1.034 penyewaan`, lalu diikuti dengan `Musim Gugur (Autumn)` dengan penyewaan sekitar `924 penyewaan` lalu selanjutnya yaitu `Musim Semi (Spring)` dengan penyewaan sepeda sekitar `746 sepeda` lalu penyewaan sepeda tersendah yaitu itu pada `Musim Salju (Winter)` dengan penyewaan sepeda sekitar `226 sepeda`.

Menggunakan pengujian One-Way ANOVA didapatkan hasil bahwa F-Statistic adalah `875.60` dan p-value < 0.001. Ini berarti bahwa `H0 ditolak` karena p-value lebih kecil daripada 0.05.

Sehingga dapat disimpulkan bahwa terdapat perbedaan rata - rata jumlah penyewaan sepeda yang signifikan antar musim. Jadi penyewaan  sepeda tertinggi pada `Musim Panas (Summer)` dan penyewaan sepeda tersendah pada `Musim Salju (Winter)`.

### 6. Perbandingan jumlah distribusi jumlah penyewaan sepeda pada Hari dengan Hujan / Salju dan Hari tanpa Hujan / Salju

#### Visualisasi menggunakan boxplot

````python
# Mengelompokkan data harian
data_harian = (
    df_functioning
    .groupby("Date")
    .agg({
        "Rented Bike Count" : "sum",
        "Rainfall(mm)": "sum",
        "Snowfall (cm)": "sum"
    })
    .reset_index()
)

data_harian
````

````python
# Membuat kategori kondisi cuaca
import numpy as np

data_harian["Kondisi Cuaca"] = np.where(
    (data_harian["Snowfall (cm)"] > 0) |
    (data_harian["Rainfall(mm)"] > 0),
    "Hujan / Salju",
    "Tanpa Hujan / Salju"
)

data_harian
````

````python
# Visualisasi menggunakan Boxplot
plt.figure(figsize = (10, 6))

sns.boxplot(
    data = data_harian,
    x = "Kondisi Cuaca",
    y = "Rented Bike Count",
    order = ["Tanpa Hujan / Salju", "Hujan / Salju"]
)

plt.title("Distribusi Penyewaan Sepeda Berdasarkan Kondisi Cuaca")
plt.xlabel("Kondisi Cuaca")
plt.ylabel("Total Penyewaan Sepeda per Hari")
plt.show()
````
